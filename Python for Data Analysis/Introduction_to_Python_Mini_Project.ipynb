### Задание 1

```python
state = True
while state:
    file_name = input("Please, enter the name of the .txt file: ")
    file_name = file_name + ".txt"
    try:
        with open(file_name, 'r') as users:
            for row in users.readlines():
                print(row)
            state = False
    except FileNotFoundError:
        print("The file that you would like to read is not found.")
    except:
        print("Some error occured.")
```

### Задание 2

```python
class Users:

    current_year = 2023
    
    def __init__(self, name, surname, birth_year, email):
        self.name = name
        self.surname = surname
        self.birth_year = birth_year
        self.email = email

    def __str__(self):
        return f"User name: {self.surname} {self.name}"

    def getAge(self):
        return self.current_year - self.birth_year
```

### Задание 3

```python
users_list = []
with open(file_name, 'r') as file:
    for row in file.readlines():
        user_attributes = row.split(',')
        new_user = Users(user_attributes[1], user_attributes[2], int(user_attributes[-1][:4]), user_attributes[3])
        users_list.append(new_user)

for user in users_list:
    print(user)
```

### Задание 4

```python
def filter(users, low = 30, high = 50):
    filtered_users = []
    age = 0
    for user in users:
        age = user.getAge()
        if age >= low and age <= high:
           filtered_users.append(f'{user.email}, {age}')
    return filtered_users

for user in filter(users_list):
    print(user)
```
