```python
df.drop('LikePresentation', axis = 1, inplace = True)
df
```

```python
import re

cols = list(df.columns)

for i in range(len(cols)):
    cols[i] = " ".join(re.split('(?<=.)(?=[A-Z])', cols[i]))

df.columns = cols
df
```

```python
df.drop_duplicates(inplace = True)
df.dropna(inplace = True)
df.shape
```
